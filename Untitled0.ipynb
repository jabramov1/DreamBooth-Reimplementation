```python
%%bash
set -e

cd "/content/drive/MyDrive/FinalProjectDL"

echo "Current folder:"
pwd
echo

echo "Files here:"
ls
echo

# Make sure this folder is a git repo
if [ ! -d ".git" ]; then
  git init
fi

# Set your Git identity
git config user.email "joseph1abramov@gmail.com"
git config user.name "Joseph Abramov"

# Set GitHub remote cleanly
git remote remove origin 2>/dev/null || true
git remote add origin https://github.com/jabramov1/DreamBooth-Reimplementation.git

# IMPORTANT:
# Remove nested .git folders inside subfolders so Git uploads their contents normally.
# This does NOT delete your actual files/images/results, only nested git metadata.
find . -mindepth 2 -type d -name ".git" -print -exec rm -rf {} +

# Unstage anything from previous attempts
git reset || true

# Add everything currently in the cleaned Drive folder
git add -A

echo
echo "What will be committed:"
git status
echo

# Commit everything
git commit -m "Upload full DreamBooth project"

# Rename branch to main
git branch -M main

# Push to GitHub
git push -u origin main
```
